Implement a function that takes a 1D NumPy array of integers and returns an array containing elements that are multiples of 5 and leave a remainder of 1 when divided by 2.

## Solution

````python
# 1. Import libraries
import numpy as np

# 2. Define a 1D NumPy array
a = np.array([1, 5, 10, 3, 4, 25, 30])

# 3. The function of the h.w that I will work on
def homework(a):
  # getting the numbers in the array that are from the multiples of 5 and when divided by 2 the remainder is 1 by NumPy
    my_result = None
    my_result = a[(a % 5 == 0) & (a % 2 == 1)]
    # we made a filter that grts the multiples of 5 and that are when divided by 2 the remainder is 1 and it acts like an if condition
    return my_result

# Run this cell after completing homework().
your_answer = homework(a)
print("Your answer:", your_answer)

# case 2
case_2_input = np.array([11, 15, 20, 21, 35, 40, 45])
case_2_answer = homework(case_2_input)
print("Your answer:", case_2_answer)

# case 3
case_3_input = np.array([2, 4, 6, 8])
case_3_answer = homework(case_3_input)
print("Your answer:", case_3_answer)


````
