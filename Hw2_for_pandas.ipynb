### Download and prepare the dataset

Then run the dataset cell. It downloads the MyAnimeList data, removes rows with
an unknown `Score`, and converts `Score` to a numeric column. Do not include
this download or loading code inside `homework()`.


````python
import numpy as np
import pandas as pd
from pandas import DataFrame

# Libraries for retrieving data from the web and handling zip files
import requests, zipfile
from io import StringIO
import io
````

````python
# Specify the url with data
url = 'https://github.com/Hernan4444/MyAnimeList-Database/archive/refs/heads/master.zip'

# Acquire data from the url
r = requests.get(url, stream=True)

# read and extract the zipfile
z = zipfile.ZipFile(io.BytesIO(r.content))
z.extractall()

# Load and clean the data
anime_data = pd.read_csv('MyAnimeList-Database-master/data/anime.csv')
anime_data_extracted = anime_data[anime_data['Score'] != 'Unknown'].copy()
anime_data_extracted['Score'] = pd.to_numeric(anime_data_extracted['Score'])
````

````python
# exploring the anime_data_extracted
print(f"The 1st 5 rows -> \n{anime_data_extracted.head()}")
print()
print(f"The last 5 rows -> \n{anime_data_extracted.tail()}")
print()
print(f"The shape of the data -> {anime_data_extracted.shape}")
print()
print(f"The columns of the data -> \n{anime_data_extracted.columns.to_list()}")
print()
print(f"The data types of the columns -> \n{anime_data_extracted.dtypes}")
print()
````

````python
# the H.W. function
def homework(anime_data_extracted):
    # getting the mean score for each type arranged in descending order from anime_data_extracted
    result = anime_data_extracted.groupby("Type")["Score"].mean().sort_values(ascending=False)
    return result

# getting the result printed
your_answer = homework(anime_data_extracted)
display(your_answer)

# the cases
case_2_data = pd.DataFrame({
    "Type": ["TV", "TV", "Movie", "Movie", "OVA"],
    "Score": [8.0, 6.0, 9.0, 7.0, 5.0],
})
case_2_answer = homework(case_2_data)
display(case_2_answer)
````
